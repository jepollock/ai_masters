# Question 1 Rejection Sampling


```python
import numpy as np
from scipy.stats import norm
from scipy.constants import pi

def p(x):
    return np.e**(-(x**2)/2) * (2 + np.sin(3 * x + 1))

def q(x):
    return np.e**(-(x**2)/2) * 3 * np.sqrt(2 * pi)

measurements = []
while len(measurements) < 10000000:
    num_samples = 1000
    x_samples = norm.rvs(loc=0, scale=1.0, size=num_samples)
    y_samples = np.random.uniform(0, 1, num_samples)
    for x, y in zip(x_samples, y_samples):
        # print(f"x = {x}, {y}")
        # print(f"p(x) = {p(x)}")
        # print(f"q(x) = {q(x) * y}")
        if q(x)*y <= p(x):
            measurements.append(x ** 2 * p(x))

expectation = np.mean(measurements)

print(f"expectation: {expectation}")
```

# About "flat" priors: Beta Distribution

```python
import numpy as np
import matplotlib.pyplot as plt

samples = np.random.beta(a=1, b=1, size=50000)
plt.figure(1)
plt.hist(samples, bins=30)
log_odds = np.log(samples/(1-samples))
plt.figure(2)
plt.hist(log_odds, bins=30)


samples_1_2 = np.random.beta(a=0.5, b=0.5, size=50000)
plt.figure(3)
plt.hist(samples_1_2, bins=30)
log_odds_1_2 = np.log(samples_1_2/(1-samples_1_2))
plt.figure(4)
plt.hist(log_odds_1_2, bins=30)

samples_1_1_coin = np.random.beta(a=4, b=2, size=50000)
samples_5_5_coin = np.random.beta(a=3.5, b=1.5, size=50000)

plt.figure(5)
plt.hist(samples_1_1_coin, bins=30, color="green")
plt.hist(samples_5_5_coin, bins=30, color="blue")
```
